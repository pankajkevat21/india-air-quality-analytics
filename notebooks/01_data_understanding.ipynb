```python
# Core libraries for data manipulation
import pandas as pd
import numpy as np

# Display settings so we can see all columns in outputs
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

# Suppress unnecessary warnings for now
import warnings
warnings.filterwarnings('ignore')

print("Libraries loaded ✅")
```

```python
# Load CSV from the raw data folder
df = pd.read_csv('../data/raw/city_day.csv')

# Confirm it loaded
print("Dataset loaded successfully ✅")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
# Look at the first 5 rows to see actual values
df.head()
```

```python
# 1. Shape
print(f"Shape: {df.shape}")

# 2. All column names
print("\nColumns:")
print(df.columns.tolist())

# 3. Data types + non-null counts (this is gold)
print("\nData Info:")
df.info()
```

```python
# Describe gives min, max, mean, quartiles for numeric columns
df.describe().T   # .T transposes for easy reading
```

```python
# Missing value count + percentage, sorted high to low
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
})

# Show only columns that HAVE missing values
missing_df[missing_df['Missing Count'] > 0]
```

```python
# How many unique cities?
print(f"Unique cities: {df['City'].nunique()}")
print("\nCity list:")
print(sorted(df['City'].unique()))

# Date range (temporarily convert for inspection)
df['Date'] = pd.to_datetime(df['Date'])
print(f"\nDate range: {df['Date'].min()} → {df['Date'].max()}")
print(f"Total days covered: {(df['Date'].max() - df['Date'].min()).days} days")
```

```python
# See what AQI_Bucket categories exist
print("AQI_Bucket value counts:")
print(df['AQI_Bucket'].value_counts(dropna=False))
```

```python
# 1. Any fully-duplicated rows?
print(f"Duplicate rows: {df.duplicated().sum()}")

# 2. Are there (City, Date) duplicates? (should be unique)
print(f"Duplicate (City, Date) pairs: {df.duplicated(subset=['City','Date']).sum()}")

# 3. Sanity check AQI: any negative values?
print(f"Negative AQI rows: {(df['AQI'] < 0).sum()}")

# 4. What's the max AQI row look like?
print("\nRow with max AQI:")
print(df.loc[df['AQI'].idxmax()][['City', 'Date', 'PM2.5', 'PM10', 'AQI', 'AQI_Bucket']])
```
